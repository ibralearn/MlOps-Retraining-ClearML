```python
# Configuration sans secret dans le notebook
import os
try:
    from dotenv import load_dotenv, find_dotenv
    env_file = find_dotenv(usecwd=True)
    if env_file:
        load_dotenv(env_file)
    else:
        load_dotenv()
except ImportError:
    pass

required = ("CLEARML_API_ACCESS_KEY", "CLEARML_API_SECRET_KEY", "CLEARML_API_HOST")
missing = [name for name in required if not os.getenv(name)]
if missing:
    raise RuntimeError("Configurez .env avant de lancer l'agent : " + ", ".join(missing))
print("Configuration ClearML chargee depuis l'environnement.")
```

```python
# Agent qui execute les etapes prepare_data, train_model et register_model.
import subprocess

QUEUE = "default"
subprocess.run(
    ["clearml-agent", "daemon", "--queue", QUEUE, "--create-queue", "--foreground"],
    check=True,
)
```
